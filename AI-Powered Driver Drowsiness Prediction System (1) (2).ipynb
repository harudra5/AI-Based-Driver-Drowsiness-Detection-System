## 2. What is Driver Drowsiness Detection (Technical Summary)

It is a **computer vision and AI-based safety system** that monitors:

* Eye closure (Eye Aspect Ratio – EAR)
* Blink duration
* Yawning frequency (Mouth Aspect Ratio – MAR)

Using:

* Camera input
* Facial landmark detection (MediaPipe / Dlib / OpenCV)
* Machine Learning or Deep Learning models

The system triggers:

* **Audio alerts**
* **Seat vibration**
* **Dashboard warnings**

```python
def left_eye_aspect_ratio(landmarks):
    p1 = landmarks[33].x, landmarks[33].y
    p2 = landmarks[133].x, landmarks[133].y

    a = math.dist(p1, p2)

    q1 = landmarks[160].x, landmarks[160].y
    q2 = landmarks[144].x, landmarks[144].y

    b = math.dist(q1,q2)
    
    r1 = landmarks[158].x, landmarks[158].y
    r2 = landmarks[153].x, landmarks[153].y

    c = math.dist(r1,r2)

    left_rear = (b+c)/(2*a)
    return left_rear
```

```python
def right_eye_aspect_ratio(landmarks):
    p1 = landmarks[362].x, landmarks[362].y
    p2 = landmarks[263].x, landmarks[263].y

    a = math.dist(p1, p2)

    q1 = landmarks[385].x, landmarks[385].y
    q2 = landmarks[380].x, landmarks[380].y

    b = math.dist(q1,q2)
    
    r1 = landmarks[387].x, landmarks[387].y
    r2 = landmarks[373].x, landmarks[373].y

    c = math.dist(r1,r2)

    right_rear = (b+c)/(2*a)
    return right_rear
```

```python
def mouth_aspect_ratio(landmarks):
    p1 = landmarks[78].x, landmarks[78].y
    p2 = landmarks[308].x, landmarks[308].y

    a = math.dist(p1, p2)

    q1 = landmarks[82].x, landmarks[82].y
    q2 = landmarks[87].x, landmarks[87].y

    b = math.dist(q1,q2)
    
    r1 = landmarks[312].x, landmarks[312].y
    r2 = landmarks[317].x, landmarks[317].y

    c = math.dist(r1,r2)

    mouth = (b+c)/(2*a)
    return mouth
```

```python
import time
import winsound
```

```python
video = cv2.VideoCapture(0)

# threshold for Eyes

EAR_THRESHOLD = 0.20 # eye close  # sleeping
EAR_COUNTER = 0
EAR_FRAME = 15

# threshold for MOUTH

MOUTH_THRESHOLD = 0.20 # mouth open  # yawing
MOUTH_COUNTER = 0
MOUTH_FRAME = 15

while True:
    success, frame = video.read()

    if not success:
        break

    image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB) # converting the image to mediapipe compactable way

    face = mp.Image(image_format = mp.ImageFormat.SRGB, data=image) # take the RGB img and convert it into mediapipe version/compatable way

    time_frame = int(time.time() * 1000) 
    result = detector.detect_for_video(image=face, timestamp_ms=time_frame)

    if result.face_landmarks:
        landmarks = result.face_landmarks[0]  # to extract the normalized face landmarks

        left_rear = left_eye_aspect_ratio(landmarks) # calling the function
        right_rear = right_eye_aspect_ratio(landmarks)

        avg_ear = (left_rear + right_rear) / 2 # if driver has only one eye it is going to say drwzy

        mouth = mouth_aspect_ratio(landmarks)

        cv2.putText(frame, text=f'Ear:{round(avg_ear, 3)}', org=(20,20),fontScale=1, fontFace=cv2.FONT_HERSHEY_TRIPLEX, color=(0,255,0), thickness=1)
        cv2.putText(frame, text=f'Mouth:{round(mouth, 3)}', org=(20,50),fontScale=1, fontFace=cv2.FONT_HERSHEY_TRIPLEX, color=(0,255,0), thickness=1)

        if avg_ear < EAR_THRESHOLD:  # Sleeping
            EAR_COUNTER += 1
        else:
            EAR_COUNTER = 0

        if mouth > MOUTH_THRESHOLD:  # Sleeping
            MOUTH_COUNTER += 1
        else:
            MOUTH_COUNTER = 0

        if (EAR_COUNTER > EAR_FRAME) or (MOUTH_COUNTER > MOUTH_FRAME): # whether you are sleeping for 1 sec or yawing for 1 se
            cv2.putText(frame, text='You are Drowsy! Take a Break!', org=(50,100),fontScale=1, fontFace=cv2.FONT_HERSHEY_TRIPLEX, color=(0,155,0), thickness=2)
            winsound.Beep(3000,1000) #frequency of soundand no of milli seconds

    cv2.imshow('webcam', frame)
    if cv2.waitKey(1) == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
